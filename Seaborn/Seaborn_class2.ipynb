# Seaborn – Class 2: Categorical Plots


## What Are Categorical Plots?

Categorical plots visualize data where at least one variable is **categorical**  
(e.g., day of week, gender, class).

Seaborn gives several easy options to:
- Compare **counts** or **averages** across categories
- Understand spread & distribution


## barplot() vs countplot()

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
tips = sns.load_dataset('tips')
tips
tips.head()
```

```python
#in tips data many categorical dat are there , sex ,day , time and size , so in  general data milte hi hain categorical form mein
sns.barplot(x="day",y="total_bill",data=tips) #pahli chiz categorical variable and second is continuous variable jiske basis pe counting hote hain and uske basis pe bar create hota hai
#sns.barplot(x=tips["day"],y=tips["total_bill"]) can also write in this form # so averga value dikhayega , ye graph mein niche candles show karta hai variance ko
#average on the basis of category
```

```python
sns.countplot(x=tips['day'])
```

```python
tips['day'].value_counts()
```

## boxplot() & violinplot()

```python
sns.boxplot(x=tips['day'],y=tips['total_bill']) #day categorical and total_bill , and can see here that oulier is only one incase of Fri
```

```python
sns.violinplot(x=tips['day'],y=tips['total_bill']) #so violinplot -> shows boxplot and also distribution
```

## stripplot() & swarmplot()

```python
sns.stripplot(x=tips['day'],y=tips['total_bill']) #can say scatter kar diya aand dikha raha hai #and kosis karta hai overlap na ho but ho hi jata hai so we can see at swramplot
```

```python
sns.swarmplot(x=tips['day'],y=tips['total_bill'])
```

```python
sns.violinplot(x=tips['day'],y=tips['total_bill'])
sns.swarmplot(x=tips['day'],y=tips['total_bill']) # and can see we can show both also together by overlap but pyara ban rah ahai
```

| Plot        | Best for |
|-------------|-----------|
| **barplot** | Compare average values across categories |
| **countplot** | Show how many items fall in each category |
| **boxplot** | See median, quartiles, and outliers |
| **violinplot** | Same as boxplot + shows distribution |
| **stripplot** | View every observation (good for small data) |
| **swarmplot** | Like stripplot but avoids overlap |
