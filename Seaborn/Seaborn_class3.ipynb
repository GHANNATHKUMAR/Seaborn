# Seaborn – Class 3: Distribution Plots

## What Are Distribution Plots?

Distribution plots show how data values are **spread or concentrated**.

Seaborn offers:
- **`histplot()`** → modern histogram (with optional KDE)
- **`kdeplot()`** → smooth curve of distribution
- **`rugplot()`** → small marks showing individual values  
*(Older `distplot()` is now replaced by `histplot()`.)*


## histplot() – Basic Histogram

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
tips=sns.load_dataset('tips')
tips.head()
```

```python
sns.histplot(tips['total_bill']) #total_bill is a continuous data
```

```python
sns.histplot(tips['total_bill'],bins=20)
```

```python
sns.histplot(tips['total_bill'],bins=50)
```

```python
sns.histplot(tips['total_bill'],bins=50,kde=True)  #kernel density estimator , #agar hamara distribution skkeweness hai aur agar koi value missing hai toh ham use karenge median , agar mera distribution normal hai toh ham use karenge mean
```

```python
sns.histplot(tips['total_bill'],bins=50,kde=True,color='red')
```

```python
plt.title("Histogram of Total bills")
```

## kdeplot() – Kernel Density Estimate

```python
sns.kdeplot(tips['total_bill']) #toh yahan sirf #kernel denstity estimator banega aur achhe se
```

```python
sns.kdeplot(tips['total_bill'],fill=True,color='green')
```

## distplot() (Legacy) vs histplot()

> `distplot()` is deprecated.
> Use `histplot()` or `kdeplot()` instead.
so some functions deplicate bhi hote hain , toh vapas se hamko documnetation me jana padt ahai aur dekhna hota hai ki currently is function ke jaagh aur kya use ho raha hai

## rugplot() – Small Ticks for Observations

```python
sns.rugplot(x=tips['total_bill'],height=0.2,color='red') #for small data it is better to see , rug plots clearly show where data points are concentrated or sparse
```

## Multiple Distributions with hue

```python
sns.histplot(x=tips['total_bill'],hue=tips['sex'],kde=True,bins=20)
```

```python
sns.histplot(x=tips['total_bill'],hue=tips['sex'],kde=True,bins=20,palette="Set2")
```

Key takeaways

histplot() → flexible histogram (can combine with KDE).

kdeplot() → smooth density curve.

rugplot() → quick look at raw values.

distplot() is deprecated → prefer histplot().

Use hue to compare groups in the same distribution plot.
