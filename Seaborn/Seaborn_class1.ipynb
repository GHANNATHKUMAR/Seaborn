## What is Seaborn?

Seaborn is a **high-level data visualization library** built on top of Matplotlib.  
It provides:
- Beautiful, modern default styles  
- Concise syntax (less code for complex visuals)  
- Integration with Pandas DataFrames  

Perfect for quick, attractive plots while keeping full Matplotlib power underneath.


###  What Are Relational Plots?

Relational plots are used to **show the relationship between two variables**.

- They focus on how one variable changes with respect to another.
- Typically used for continuous (numeric) data, but can also mix with categories.

In Seaborn, the main relational plots are:

1. **`scatterplot()`** → displays individual data points  
2. **`lineplot()`** → shows a continuous trend or summary line

> Both functions belong to Seaborn’s *relational* family.

### Scatter Plot with scatterplot()

```python
sns.scatterplot(x=tips['total_bill'],y=tips['tip']) #here only in one line no need to assign a figure etc , so main things for plot and for designing we can use plt
plt.title("Scatter plots between total and tip ")
#so we can see toatal bill increase with tip
#also we can connect with relation becuase it is relational plots , so we can also undestand corelation and covariance how increase
```

```python
tips['total_bill'].corr(tips['tip']) #means 67 % chance is increase tips when bill increase
```

### Line Plot with lineplot()

```python
sns.lineplot(x=tips['day'],y=tips['tip']) #only 7 days
```

```python
sns.lineplot(x=tips['day'],y=tips['tip'],estimator='mean',marker='o') #estimator we can use in lineplot
```

### Styling & Color Palettes

```python
sns.set_style('darkgrid') #options : white , dark ,whitegrid , whitegrid ,ticks
sns.set_palette("bright") #others : deep , bright , dark , muted , colorblind
sns.scatterplot(x=tips['total_bill'],y=tips['tip'])
```

```python
sns.set_style('darkgrid')
sns.set_palette("bright")
sns.scatterplot(x=tips['total_bill'],y=tips['tip'],hue=tips['sex'])
```

```python
sns.set_style('ticks')
sns.set_palette("dark")
sns.scatterplot(x=tips['total_bill'],y=tips['tip'],hue=tips['sex'])
```
