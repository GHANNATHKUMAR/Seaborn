## Why Matrix / Grid Plots?

Matrix or grid-style plots are perfect for:
- Showing correlations between variables  
- Visualizing pivot tables  
- Discovering patterns in large datasets


### Heatmaps (heatmap())

```python
import seaborn as sns
import matplotlib.pyplot as plt
tips=sns.load_dataset('tips')
tips.head()
```

```python
tips.corr(numeric_only=True) #we can aslo show its on plots
```

```python
sns.heatmap(tips.corr(numeric_only=True))
```

```python
sns.heatmap(tips.corr(numeric_only=True),annot=True) #isse ye bhi pata chalega ki kaun si kis se kitne corelated hain , # it is important to create model how much corelated , and it works only on numeric columns
```

### Cluster Map (clustermap())

```python
#iska use tab samajh aayeg ajab aap clustering in ml
```

```python
sns.clustermap(tips.corr(numeric_only=True),annot=True)
```

```python
iris=sns.load_dataset('iris')
sns.clustermap(iris.corr(numeric_only=True),annot=True) #same chiz hota hai but batata hai ki kaun kis se jayda connected hai and kaun kis se kam connected hai
```

### Styling & Themes

```python
sns.set_style('darkgrid')
sns.set_palette('Set2')
sns.scatterplot(x=tips['total_bill'],y=tips['tip'],hue=tips['sex'])
```

```python
sns.scatterplot(x=tips['total_bill'],y=tips['tip'],hue=tips['sex'],s=80)  #s is size
```
