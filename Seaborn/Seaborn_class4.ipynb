# Seaborn – Class 4: Regression & Mixed Plots

## Regression & Mixed Plots

Seaborn provides plots that combine **relationships** and **statistical models** or mix multiple plot types.

- `regplot()` / `lmplot()` → scatter + regression line  
- `jointplot()` → scatter with hist/KDE  
- `pairplot()` → multiple relationships in one go


# regplot() – Scatter + Regression Line

```python
import seaborn as sns
import matplotlib.pyplot as plt
tips=sns.load_dataset('tips')
tips.head()
```

```python
sns.regplot(x=tips['total_bill'],y=tips['tip'],color='green') #regression plot , a regression line
plt.title("Regression plot")
```

# lmplot() – Regression with Facets

```python
sns.lmplot(x='total_bill',y='tip',data=tips,height=5,aspect=1.2)
```

```python
sns.lmplot(x='total_bill',hue='sex',y='tip',data=tips,height=5,aspect=1.2) #so ek regression line male ke liye and ek regression ine hai female ke liye
```

```python
#so regression plot mein hue nahi dekh sakte and lmplot mein hue de sakte hain
```

# jointplot() – Scatter + Histogram / KDE

```python
sns.jointplot(x=tips['total_bill'],y=tips['tip'])
```

```python
sns.jointplot(x=tips['total_bill'],y=tips['tip'],kind='scatter')
```

```python
sns.jointplot(x=tips['total_bill'],y=tips['tip'],kind='hex') #by default scatter and here we have chosed hex
```

```python
sns.jointplot(x=tips['total_bill'],y=tips['tip'],kind='kde')
```

```python
sns.jointplot(x=tips['total_bill'],y=tips['tip'],kind='kde',fill=True)
```

# pairplot() – Relationships Between Many Variables

```python
iris=sns.load_dataset('iris')
iris
```

```python
sns.pairplot(iris)
```

```python
sns.pairplot(iris,hue='species')
```
